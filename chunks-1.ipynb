### Notes on NLP

- **Morphology:** The study of word formation
- **Phonemes:** The smallest sound you make that distinguishes words e.g `cat`, `mat`
- **Morpheme:** Combination of multiple phonemes. It is the smallest unit of meaning in a word e.g `un-believe-able` :: `un-` means not/negation, `-believ-` is the root, and `-able` indicates ability. It is used to interpret the meaning of words and their relations
- **Lexemes:** Variations of Morphemes e.g `speak`, `spoke`, `spoken`. They are key for semantic analysis and text normalisation
- **Syntax:** Checks the rules of the language to see if it makes sense
- **Semantics:** Refers to the meaning of words and sentences within context
- **Stemming:** Is the process of reducing words to their base or root form, such as "running" to "run," to normalize words and improve text processing efficiency. 
- **Lemmatization:** Reduces words to their base form (lemma) considering context and meaning, unlike stemming, making it more accurate for text processing.

```python
# Import the library

import nltk;
```

```python
text = 'Ore moved to Paris to experience culture but then realised he doesn\'t like France. He is currently in Scotland, where the weather is much depressing, doing a course on NLP'
text
```

```python
nltk.download('punkt');

nltk.download('punkt_tab');
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.tokenize import word_tokenize;
```

```python
tokens = word_tokenize(text);
print(tokens);
```

```python
# Stemming 

from nltk.stem import PorterStemmer;

stemmer = PorterStemmer();
stemmedWords = [stemmer.stem(token) for token in tokens];
print(stemmedWords);
```

```python
# Part-of-speech Tagging

from nltk import pos_tag;

textPosTag = pos_tag(tokens);

print(textPosTag);
```

```python
# Named Entity Recognition

nltk.download("maxent_ne_chunker");
nltk.download('maxent_ne_chunker_tab')
nltk.download("words");
entities = nltk.ne_chunk(textPosTag);
print(entities);
```
